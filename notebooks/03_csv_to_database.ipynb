```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text("SELECT version();"))
    print(result.fetchone()[0])
```

```python
table_name = "ipl_ball_by_ball"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=10000
)

print("Dataset uploaded successfully!")
```
